```python
import json
from etl.config import SessionLocal
from etl.load import ProductModel
from backend.analytics import get_pricing_intelligence

# Busca um product_id real do banco
session = SessionLocal()
sample_product = session.query(ProductModel).first()
session.close()

if not sample_product:
    print("Nenhum produto encontrado no banco. Rode o pipeline de load primeiro!")
else:
    target_id = sample_product.product_id
    print(f"Testando analytics para o produto: {sample_product.canonical_name} (ID: {target_id})\n")

    # Executa a inteligência analítica
    intelligence_payload = get_pricing_intelligence(target_id)

    # Exibe o resultado formatado
    print(json.dumps(intelligence_payload, indent=4, ensure_ascii=False))
```
